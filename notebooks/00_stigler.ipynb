```python
from pulp import LpMinimize, LpProblem, LpStatus, LpVariable, lpSum

# 1. Os dicionários (agora com o sódio para a S0-07)
precos = {"Arroz": 5.0, "Feijao": 7.0, "Frango": 15.0, "Alface": 3.0, "Ovo": 8.0}
calorias = {"Arroz": 130, "Feijao": 90, "Frango": 165, "Alface": 15, "Ovo": 155}
proteinas = {"Arroz": 2.5, "Feijao": 5.0, "Frango": 31.0, "Alface": 1.0, "Ovo": 13.0}
sodio = {"Arroz": 1, "Feijao": 2, "Frango": 50, "Alface": 5, "Ovo": 150}

alimentos = list(precos.keys())

# 2. Problema
prob = LpProblem(name="Dieta_Stigler_Simples", sense=LpMinimize)

# 3. Variáveis
x = LpVariable.dicts("Quantidade", alimentos, lowBound=0, cat="Continuous")

# 4. Função Objetivo (sem sinal de comparação)
prob += lpSum([precos[i] * x[i] for i in alimentos]), "Custo_Total"

# 5. Restrições da S0-06
prob += lpSum([calorias[i] * x[i] for i in alimentos]) >= 1800, "Calorias_Minimas"
prob += lpSum([calorias[i] * x[i] for i in alimentos]) <= 2200, "Calorias_Maximas"
prob += lpSum([proteinas[i] * x[i] for i in alimentos]) >= 50, "Proteina_Minima"

# 6. Experimentos da S0-07
# Experimento 1: Teto folgado (o custo vai subir porque ovo tem muito sódio)
# prob += lpSum([sodio[i] * x[i] for i in alimentos]) <= 100, "Sodio_Maximo"

# Experimento 2: Teto apertado (vai dar Infeasible)[cite: 2]
# prob += lpSum([sodio[i] * x[i] for i in alimentos]) <= 10, "Sodio_Maximo_Apertado"

# 7. Resolver e Imprimir
prob.solve()

print(f"Status: {LpStatus[prob.status]}")
print(f"Custo Mínimo: R$ {prob.objective.value():.2f}")

print("\nCardápio Escolhido:")
for i in alimentos:
    if x[i].varValue > 0:
        print(f"- {i}: {x[i].varValue:.2f} porções")
```
